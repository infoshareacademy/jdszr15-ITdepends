```python
from sympy import *
from sympy.stats import *
from IPython.display import display
import matplotlib.pyplot as plt

init_printing()
```

**Definicja jednostajnego rozkładu dyskretnego (rzut kostką do gry)**

```python
x = Symbol("x")
X = DiscreteUniform('X', [1,2,3,4,5,6]) 

display( density(X) )

p = density(X)(x)
display(p)

EX = E(X) # wartość oczekiwana
display(EX)

DX = variance(X) # wariancja
display(DX)

pr = P(X<2.5) # prawdopodobieństwo, że  X < 2.5
display( pr )



t = list(range(1,7))
tp = list(map(lambda v: float(p.subs(x,v)),t))
fig, ax = plt.subplots(1, 1)
ax.plot(t, tp, 'ro', ms=12, mec='r')
ax.vlines(t,  0, tp, colors='r', lw=1)
plt.show()
```

**Możliwe jest też utworzenie dowolnej zmiennej losowej dyskretnej:**

```python
d = {0: 0.1, 1: .2, 2: .3, 3: .4}
X = FiniteRV('X', d)

display( density(X) )

p = density(X)(x)
display(p)

EX = E(X)
display(EX)

DX = variance(X)
display(DX)

display(P(X<2.5))

t = list(range(0,len(d)+1))
tp = list(map(lambda v: float(P(Eq(X,v))),t))
fig, ax = plt.subplots(1, 1)
ax.plot(t, tp, 'ro', ms=12, mec='r')
ax.vlines(t,  0, tp, colors='r', lw=1)
plt.show()
```

```python
p1 = Symbol("p1",nonnegative=True)
p2 = Symbol("p2",nonnegative=True)
p3 = Symbol("p3",nonnegative=True)
d = {0: p1, 1: p2, 2: p3} # założone prawdopodobieństwa

x = Symbol("x")
X = FiniteRV('X', d)
display(density(X))

p = density(X)(x)
display(p)

F = cdf(X)
display(F)

display( density(X) )

p = density(X)(x)
display(p)

EX = simplify(E(X)) 
display(EX)

DX = simplify(variance(X)) 
display(DX)
```

**Przykład rozkładu Poissona**

```python
l = Symbol("lambda", positive=True)

x = Symbol("x",positive=True)
X = Poisson("X", l)

p = density(X)(x)
display(p)

EX = E(X) 
display(EX)

DX = variance(X) 
display(DX)
```
