```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt
```

#### Definicja rozkładu i losowanie wartości pochodzących z niego

```python
rozklad1 = st.norm(12,0.5)
x = rozklad1.rvs(size=250)

plt.hist(x,bins="auto",edgecolor="black",density=True)
pass
```

```python
rozklad_do_est = st.norm # określenie klasy rozkładu użytego do estymacji
parametry_est = rozklad_do_est.fit(x)
parametry_est
```

```python
rozklad_estymowany = rozklad_do_est(*parametry_est)
rozklad_estymowany
```

```python
plt.hist(x,bins="auto",edgecolor="black",density=True)
t = np.linspace(min(x),max(x),1000)
plt.plot(t,rozklad_estymowany.pdf(t))
pass
```
