W zeszycie przedstawiono podstawy statystyki opisowej na przykładzie **bardzo "kłopotliwego" zbioru danych**

```python
import numpy as np
import scipy as sp
import scipy.stats as ssp
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
N1 = 100
N2 = 100
X = ssp.norm(0,0.1)
Y = ssp.norm(10,0.1)
x = np.concatenate([X.rvs(N1),Y.rvs(N2,2)])
print(x)
```

**Wykres skrzynkowy**

```python
plt.boxplot(x,notch=True,vert=False)
pass
```

**Dlatego zawsze należy robić histogram**

```python
plt.hist(x)
pass
```

**A najlepiej histogram i boxplot na jednym wykresie**

```python
f, (ax_box, ax_hist) = plt.subplots(2, sharex=True,gridspec_kw={"height_ratios": (1, 1)})
ax_box.boxplot(x,notch=True,vert=False)
ax_box.axis('off')
ax_hist.hist(x)
```

**Funkcje wyższego poziomu do wizualizacji statystycznej w Python**

```python
import seaborn as sns
```

```python
sns.displot(x);
```

```python
sns.displot(x, kde=False, rug=True);
```
