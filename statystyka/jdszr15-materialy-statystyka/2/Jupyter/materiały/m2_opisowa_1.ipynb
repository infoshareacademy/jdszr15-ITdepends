Podstawowe elementy analizy opisowej

```python
import numpy as np
import scipy as sp
import scipy.stats as ssp
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
N1 = 1000
X = ssp.norm(0,0.1)
x = X.rvs(N1)
#print(x)
```

**Wykres skrzynkowy**

```python
plt.boxplot(x,notch=True,vert=False)
pass
```

**Histogram**

```python
plt.hist(x,edgecolor="black")
pass
```

**Histogram i boxplot na jednym wykresie**

```python
f, (ax_box, ax_hist) = plt.subplots(2, sharex=True,gridspec_kw={"height_ratios": (1, 1)})
ax_box.boxplot(x,notch=True,vert=False)
ax_box.axis('off')
ax_hist.hist(x,edgecolor="black")
pass
```
