```python
from scipy import stats as st
```

Narysować wykres rozkładu prawdopodobieństwa i dystrybuanty zmiennej losowej $X$, dla której: $P(X=0)=\frac{1}{10}$,  $P(X=1)=\frac{9}{10}$. Obliczyć jej wartość oczekiwaną, wariancję i odchylenie standardowe.

W module scipy znajduje się obiekt rv_discrete, który można wykorzystać do dwfiniowania własnego dyskretnego rozkładu prawdopodobieństwa. W celu pracy z rozkłądem wystarczy użyć wbudowanych metod np. mean obliczy wartość oczekiwaną.

```python
distribution311 = st.rv_discrete(name='Rozkład z zadania 3.1.1', values=((0,1), (1/10,9/10)))
```

```python
E = distribution311.expect() # wartość oczekiwana
E
```

```python
S = distribution311.std() # odchylenie standardowe
S
```

```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
xk=distribution311.xk
p = distribution311.pmf(xk)
F = distribution311.cdf(xk)
plt.plot(xk, p, 'ro', ms=8, mec='r')
plt.vlines(xk, 0, p, colors='r', lw=4)
plt.hlines(np.append([0],F), np.append([min(xk)-1],xk), np.append(xk,[max(xk)+1]), colors='b', lw=2)
pass
```

**Znaleźć rozkład prawdopodobieństwa, dystrybuantę, wartość oczekiwaną i wariancję zmiennej losowej $X$ – sumy oczek otrzymanej w wyniku rzutu dwiema czworościennymi kostkami do gry.**

Wygenerownie wszystkich możliwych wartości przyjmowanych przez tę zmienną losową i odpowiadających im częstości

```python
W=dict()
for i in range(1,5):
    for j in range(1,5):
        if i+j in W:
            W[i+j]+=1 # jeśli klucz już istnieje po prostu dolicz do odpowiadajacej mu wartości 1
        else:
            W[i+j]=1 # jeśli klucz nie istnieje, utwórz klucz i ustaw liczbę jego wystąpień równą 1
            
W  # wypisz zdarzenia i częstości
```

```python
P=sorted(list(W.items()), key=lambda x: x[0]) # posortowanie krotek po wartościach x

xk = list(map(lambda x:x[0],P)) # wartości przyjmowane przez zmienną losową
#określenie prawdopodobieństw
pk = list(map(lambda x:x[-1],P))
N = sum(pk)
pk = list(map(lambda x:x/N,pk)) # ostatecznie - policzone prawdopodobieństwa

distribution312 = st.rv_discrete(name='Rozkład z zadania 3.1.2', values=(xk, pk)) #utworzenie rozkładu
```

```python
E = distribution312.expect() # wartość oczekiwana
E
```

```python
V = distribution312.var() # wariancja
V
```

```python
xk=distribution312.xk
p = distribution312.pmf(xk)
F = distribution312.cdf(xk)
plt.plot(xk, p, 'ro', ms=8, mec='r')
plt.vlines(xk, 0, p, colors='r', lw=4)
plt.hlines(np.append([0],F), np.append([min(xk)-1],xk), np.append(xk,[max(xk)+1]), colors='b', lw=2)
pass
```
