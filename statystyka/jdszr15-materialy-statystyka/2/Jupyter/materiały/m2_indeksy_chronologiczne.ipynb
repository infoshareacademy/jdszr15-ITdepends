całkowitego wydobycia węgla w Polsce w latach 2010-2017
dane: coal.xlsx

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("coal.xlsx") # import danych z pliku Excela
df # wypisanie danych
```

```python
del df["No"] # Usunięcie kolumn z numeracją wierszy
df # wypisanie danych
```

```python
df["Total"] = df[df.columns[1]]+df[df.columns[2]]+df[df.columns[3]] # utworzenie kolumny "Total" 
                                                                  # sumującej wszystkie rodzaje węgla
df # wypisanie danych
```

```python
df[df.columns[1:]].describe() # Obliczenie podstawowych charakterystyk dla kolumn z pominięciem kolumny 0 [Year]
```

```python
df[df.columns[1:]].mad() # Obliczenie odchylenia przeciętnego dla kolumn
```

```python
df.plot.scatter(df.columns[0],df.columns[1]) # wykres wydobycia węgla w czasie
pass
```

```python
DIndex = 100*df["Total"][1:].to_numpy()/df["Total"][0:-1].to_numpy() # Obliczenie indeksu dynamicznego
df["DIndex"] = pd.concat([pd.DataFrame([np.NaN]),pd.DataFrame(DIndex)], ignore_index=True) #dodanie tego indeksu jako nowej kolumny
df #wypisanie danych
```
