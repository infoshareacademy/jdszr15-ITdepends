```python
from scipy import stats as st
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
class moj_rozklad(st.rv_continuous):
    "rozklad z zadania 3.2.3"
    def _pdf(self, x):
        return 2./(np.pi*(np.exp(x)+np.exp(-x)))
```

```python
rozklad = moj_rozklad(name='instancja mojego rozkladu')
```

```python
x = np.arange(-5,5,0.02)
plt.plot(x,rozklad.pdf(x),lw=3) # wykres rozkłądu gęstości
plt.plot(x,rozklad.cdf(x),"red") # wykres dystrybuanty
pass
```

```python
Q = rozklad.ppf([0.1,0.5,0.9]) # obliczenie kwantyli przy pomocy funkcji ppf
print(Q)
```
