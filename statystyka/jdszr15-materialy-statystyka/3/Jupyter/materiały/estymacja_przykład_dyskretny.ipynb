```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt
```

#### Definicja rozkładu i losowanie wartości pochodzących z niego

```python
rozklad_dyskretny_teoretyczny = st.binom(5,0.5)
x = rozklad_dyskretny_teoretyczny.rvs(size=250)

k,nk = np.unique(x,return_counts=True)
pk = nk/np.sum(nk)

plt.vlines(k,0,pk,color="red")
plt.vlines(k,0,rozklad_dyskretny_teoretyczny.pmf(k),color="blue",alpha=0.5,linewidth=4)
plt.legend(["próba","rozkład teoretyczny"])
pass
```

```python
rozklad_do_est = st.binom # określenie klasy rozkładu użytego do estymacji
bounds = [(0, 5), (0, 1)]
parametry_est = st.fit(rozklad_do_est,x,bounds)
parametry_est
```

```python
rozklad_estymowany = rozklad_do_est(*parametry_est.params)
rozklad_estymowany
```

```python
plt.vlines(k,0,rozklad_dyskretny_teoretyczny.pmf(k),color="blue",alpha=0.5,linewidth=4)
plt.vlines(k,0,pk,color="red")
plt.vlines(k,0,rozklad_estymowany.pmf(k),color="black",linestyles='dotted',linewidth=3)
plt.legend(["próba","rozkład teoretyczny","rozkład estymowany"])
pass
```
