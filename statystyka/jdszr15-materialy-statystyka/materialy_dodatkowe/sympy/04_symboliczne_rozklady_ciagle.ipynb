```python
from sympy import *
from sympy.stats import *
from IPython.display import display

init_printing()
```

**Manipulacje symboliczne**
Jeśli parametry rozkładu są symbolami, wówczas wszystkie operacje są wykonywane symbolicznie.

```python
x = Symbol("x")
mu = Symbol("mu")
s = Symbol("sigma",positive = True)
U = Normal("U",mu, s) # tworzenie instancji zmiennej losowej o rozkładzie normalnym

f = density(U)(x) # gęstosć rozkładu prawdopodobieństwa
display(f)

F = simplify( cdf(U)(x) ) # dystrybuanta
display(F)

     

EU = simplify( E(U) ) # wartość oczekiwana
display( EU )

DU = simplify( variance(U) ) #wariancja
display( DU )
```

```python
U2 = U**2 # nowa zmienna losowa, która jest kwadratem U
EU2 = simplify( E(U2) ) #określenie jej wartości oczekiwanej
display(EU2)
```

Jeśli do rozkładu wprowadzone zostaną wartości liczbowe, obliczenia są wykonywane na liczbach

```python
x = Symbol("x")
U = Normal("U",0, 1) # utworzenie zmiennej losowej o rozkładzie normalnym

f = density(U)(x) # gęstość rozkładu
display(f)

F = cdf(U)(x) # dystrybuanta
display(F)

pr = P(U>0.25) # obliczenie prawdopodobieństwa, że zmienna losowa U osiągnie wartość większą od 0.25
display(pr)
display( float(pr) )         

EU = E(U) # określenie wartości oczekiwanej
display( EU )

DU = variance(U) #wariancja
display( DU )

#make a plot
p = plot((f,(x,-4,4)),(F,(x,-4,4)),show=False,legend=True)
p[0].line_color = 'r'
p[0].label = "$f(x)$"
p[1].label = "$F(x)$"
p.xlabel = "$x$"
p.ylabel= "$f(x), F(x)$"
p.show()
```
