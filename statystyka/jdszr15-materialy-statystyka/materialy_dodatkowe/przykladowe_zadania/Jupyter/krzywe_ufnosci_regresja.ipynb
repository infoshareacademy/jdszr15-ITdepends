```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
xi = np.array([1,  2,  3,  4,  5,  6,  7])
yi = np.array([8, 13, 14, 17, 18, 20, 22])
n = len(xi)
alfa = 1-0.95
```

```python
lin_regression = st.linregress(xi,yi) # wykonanie metody najmniejszych kwadratów i obliczenie współczyników a i b
a = lin_regression.slope
b = lin_regression.intercept

print("y = {a:.4}*x + {b:.4}".format(a=a,b=b))
```

```python
plt.scatter(xi,yi)
plt.plot(xi[[0,-1]],a*xi[[0,-1]]+b,"red")
pass
```

```python
yp = a*xi+b
sr = np.sqrt(np.sum( (yi-yp)**2 ) /(n-2) )

mx = np.mean(xi)
syi =sr*np.sqrt(1/n+(xi-mx)**2/np.sum( (xi-mx)**2 ))

StudentTDistribution = st.t(n-2)
ta = StudentTDistribution.ppf([alfa/2,1-alfa/2])

ye1 = yp + ta[0]*syi
ye2 = yp + ta[1]*syi
```

```python
plt.scatter(xi,yi)
plt.plot(xi[[0,-1]],a*xi[[0,-1]]+b,"red")
plt.plot(xi,ye1,"green")
plt.plot(xi,ye2,"green")
pass
```
