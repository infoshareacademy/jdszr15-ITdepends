### W module SymPy zdefiniowane są również operacje różniczkowania i całkowania

```python
%matplotlib inline

from sympy import *
from sympy.plotting import *
init_printing()
```

```python
a = Symbol("a") # definicja symbolu matematycznego w SymPy
b = Symbol("b") # definicja symbolu matematycznego w SymPy
c = Symbol("c") # definicja symbolu matematycznego w SymPy
x = Symbol("x") # definicja symbolu matematycznego w SymPy
```

### Różniczkowanie

Obliczenie pochodnej funkcji $f(x)=ax^2+bx+c$ względem zmiennej $x$:

```python
f = a*x**2+b*x+c
diff(f,x)
```

Obliczenie pochodnej funkcji $g(x)=\frac{ax^2+bx+c}{sin(x)^a}$ względem zmiennej $x$:

```python
g = f/sin(x)**a
diff(g,x)
```

### Całkowanie

Obliczenie całki funkcji $h(x) = cos^2(x)$, tj $\int cos^2(x) dx$:

```python
h = cos(x)**2
integrate(h,x)
```

Obliczenie całki $\int_a^b\lambda e^{-\lambda x} dx$ dla $\lambda>0$

```python
l = Symbol("lambda", positive = True)
integrate(l*exp(-l*x), (x, a, b))
```
