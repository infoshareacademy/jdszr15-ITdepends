Serializacja (zrodlo: wiki)– <i> w programowaniu proces przekształcania obiektów, tj. instancji określonych klas, do postaci szeregowej, czyli w strumień bajtów, z zachowaniem aktualnego stanu obiektu. Serializowany obiekt może zostać utrwalony w pliku dyskowym, przesłany do innego procesu lub innego komputera poprzez sieć. Procesem odwrotnym do serializacji jest deserializacja. Proces ten polega na odczytaniu wcześniej zapisanego strumienia danych i odtworzeniu na tej podstawie obiektu klasy wraz z jego stanem bezpośrednio sprzed serializacji. <i/>
    

## Pickle

W Pythonie do serializacji/deserializacji danych najczęściej używany jest moduł pickle. Serializację wykonuje się z użyciem metody pickle.dumps, zaś deserializację metodą pickle.loads.

```python
# Import the Logistic Regression Module from Scikit Learn
from sklearn.linear_model import LogisticRegression

# Import the IRIS Dataset to be used in this Kernel
from sklearn.datasets import load_iris

# Load the Module to split the Dataset into Train & Test
from sklearn.model_selection import train_test_split
```

```python
# Load the data
Iris_data = load_iris()
```

```python
# Split data
Xtrain, Xtest, Ytrain, Ytest = train_test_split(Iris_data.data,
                                                Iris_data.target,
                                                test_size=0.3,
                                                random_state=4)
```

```python
# Define the Model
LR_Model = LogisticRegression(C=0.1,
                               max_iter=20,
                               fit_intercept=True,
                               n_jobs=3)

# Train the Model
LR_Model.fit(Xtrain, Ytrain)
```

```python
# Import pickle Package

import pickle
```

```python
# Save the Modle to file in the current working directory

Pkl_Filename = "Pickle_RL_Model.pkl"

with open(Pkl_Filename, 'wb') as file:
    pickle.dump(LR_Model, file)
```

```python
# Load the Model back from file
with open(Pkl_Filename, 'rb') as file:
    Pickled_LR_Model = pickle.load(file)

Pickled_LR_Model
```

```python
# Use the Reloaded Model to
# Calculate the accuracy score and predict target values

# Calculate the Score
score = Pickled_LR_Model.score(Xtest, Ytest)
# Print the Score
print("Test score: {0:.2f} %".format(100 * score))

# Predict the Labels using the reloaded Model
Ypredict = Pickled_LR_Model.predict(Xtest)

Ypredict
```

## Joblib

Moduł Joblib jest dostępny z pakietu Scikit Learn i ma zastąpić Pickle dla obiektów zawierających duże dane.


```python
# Import Joblib Module from Scikit Learn
import joblib
```

```python
joblib_file = "joblib_RL_Model.pkl"
joblib.dump(LR_Model, joblib_file)
```

```python
# Load from file

joblib_LR_model = joblib.load(joblib_file)


joblib_LR_model
```

```python
# Use the Reloaded Joblib Model to
# Calculate the accuracy score and predict target values

# Calculate the Score
score = joblib_LR_model.score(Xtest, Ytest)
# Print the Score
print("Test score: {0:.2f} %".format(100 * score))

# Predict the Labels using the reloaded Model
Ypredict = joblib_LR_model.predict(Xtest)

Ypredict
```
