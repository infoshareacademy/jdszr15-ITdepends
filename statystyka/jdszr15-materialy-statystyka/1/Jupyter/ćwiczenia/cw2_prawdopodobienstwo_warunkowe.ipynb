## Prawdopodobieństwo warunkowe

Zbiór danych ze strony [kaggle](https://www.kaggle.com/jaimevalero/developers-and-programming-languages/version/2) dotyczący użycia języków programowania i technologii używanych przez developerów.

Zadanie: Na podstawie odpowiedzi respondentów określ prawdopodibeństwotego, że developer:
 - używa języka Python
 - używa środowiska Jupyter
 - używa środowiska Jupyter jeśli wiadomo, że używa języka Python
 - używa języka Python jeśli wiadomo, że używa środowiska Jupyter

**Wczytanie i przygotowanie danych** (nie jest to bezpośrednią częścią zadania z rachunku prawdopodobieństwa)

```python
import pandas as pd #import biblioteki Pandas do łatwej manipulacji danymi
```

```python
df = pd.read_csv("python_jupyter_data.csv", index_col=0) # wczytanie pliku z danymi
df.head() # wypisanie części pliku w postaci tabeli
```

```python
df1=pd.DataFrame() #utworzenie pustego data frame (obiektu Pandas przechowujacego dane)
df1["jupyter"] = (df["jupyter notebook"]+df["jupyter"]+df["jupyter-notebook"])>0 #wybranie wszystkich rekordów, w których wspomniano termin jupyter i przypisanie im wartości 1
df1["python"] = (df["python"]+df["ipython"]+df["ipython-notebook"])>0 #wybranie wszystkich rekordów, w któryc wspomniano pythona
df1.dropna(inplace=True) # usunięcie wartości brakujących, jeśli jakieś w ogóle są
del df #usunięcie oryginalnego data frame - nie będzie już potrzebny

df1.head() #wypisanie części tabelki zawierającej dane o Pythonie i Jupyterze
```

```python
T = df1.describe() # podsumowanie utworzonej tabelki i zapisanie jej do zmiennej T
T
```

**Obliczenie potrzebnych prawdopodobieństw**

```python
N = T.loc["count"][0] # wybranie liczebności próby, tj . odpowiedzi respondentów N = 17461
Nj = sum(df1["jupyter"]) # liczba respondentów używająca technologii jupyter
Np = sum(df1["python"]) # liczba respondentów używająca pythona
Npj = sum( (df1["jupyter"]) & (df1["python"])) # liczba respondentów używająca jupytera oraz pythona

P = dict() # utworzenie słownika języka Python - wygodny do przechowywania prawdopodobieństw
P["jupyter"] = Nj / N # prawdopodobieństwo, ze respondent używa środowiska Jupyter
P["python"] = Np / N # prawdopodobieństwo, że respondent używa języka Python
P["jupyter and python"] = (Npj) / N # prawdopodobieństwo, że respondent używa jupytera oraz pythona
P["jupyter|python"] = P["jupyter and python"]/P["python"] # prawdopodobieństwo, że respondent używa jupytera jeśli używa pythona
P["python|jupyter"] = P["jupyter and python"]/P["jupyter"] # prawdopodobieństwo, że respondent używa pythona jeśli używa jupytera
P
```

**Do zadania można podejśc inaczej: Utworzyć tabelkę respondentów używających / nie używających Pythona oraz Jupytera i obliczyć odpowiednie prawdopodobieństwa. Oczywiście wyniki muszą wyjść identyczne**

```python
Npj # respondenci używający pythona oraz juptera (Python, jupyter)
Np_nj = sum( (df1["python"]==True) & (df1["jupyter"]==False)) # respondenci używający pythona ale nie jupytera (python, nie jupyter)
Nij = sum( (df1["python"]==False) & (df1["jupyter"]==True)) # respondenci używający jupytera z innym językiem niż python
Ni_nj = sum( (df1["python"]==False) & (df1["jupyter"]==False)) # respondenci nie używający pythona ani jupytera

sum([Npj,Np_nj,Nij,Ni_nj])==N # sprawdzenie czy liczba elementów jest równa liczbe obserwacji (zawsze należy to zrobić)
```
