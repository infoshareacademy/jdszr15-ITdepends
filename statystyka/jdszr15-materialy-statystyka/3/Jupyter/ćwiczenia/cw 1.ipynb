Dany jest zbiór informacji o aplikacjach Google Play. Przyjmując poziom ufności $1−\alpha=0,99$, określ przedział ufności dla średniej oceny aplikacji w kategorii Tools.

```python
import scipy.stats as st
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("googleplaystore.csv") # import danych do obiektu DataFrame
print(df.columns) # wypisanie nazw kolumn
print(df['Category'].unique()) # wypisanie unikalnych wartości z kolumny "Category"
df.head() # wypisanie kilku początkowych wierszy danych
```

**Wybór danych, dla których kategoria to TOOLS. Ponadto wybór jedynie kolumny Rating. Nadpisanie obiektu df**

```python
df = df.loc[df['Category']=="TOOLS", 'Rating']
df
```

**Czy DataFrame zawiera obserwacje NaN?**

```python
print(not df.count() == len(df) )
```

**Usunięcie wartości NaN**

```python
df = df.dropna() # usunięcie wartości nieliczbowych
print(not df.count() == len(df) )
```

```python
x = df.to_numpy()
n = len(x)
m = st.tmean(x)    
s = st.tstd(x)  
```

```python
a = 0.01

NormalDistribution = st.norm() # utworzenie instancji obiektu rozkładu normalnego standardyzowanego

ua = NormalDistribution.ppf(1-a/2) # obliczenie prawego kwantyla rozkładu normalnego standardyzowanego
CIm = np.array( [m - ua*s/np.sqrt(n), m + ua*s/np.sqrt(n)] ) # obliczenie granic przedziałów ufności

print("Przedział ufności pokrywający średnią wartości w populacji generalnej z p={}:".format(1-a))
print(CIm)
```

#### Użycie funkcji bootstrap do określenia przedziałów ufności:
W praktyce nie ma potrzeby samodzielnego określania przedziałów ufności. Biblioteka scipy.stats co prawda nie oferuje funkcji dedykowanych obliczaniu przedziałów ufności dla różnych parametrów, jednak oferuje jedną funkcję umożliwiającą obliczenie przedziałów ufności dla dowolnego parametru. Jest to funkcja [`bootstrap`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.bootstrap.html).

Poniżej wykorzystaną tę funkcję do oszacowania przedziałów ufności dla średniego ratingu.

```python
# alfa == significance_level
#1-alfa == confidence_level
#do bootstrapa dane podajemy w formie krotki
dane = (x,)
st.bootstrap(dane,st.tmean,confidence_level=1-a)
```
