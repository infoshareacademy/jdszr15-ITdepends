## Prosta statystyka symboliczna z użyciem Sympy

```python
%matplotlib inline

from sympy import *
from sympy.stats import *
from sympy.plotting import *
init_printing()
```

```python
a = Symbol("a", positive = True) # definicja symbolu matematycznego w SymPy
x = Symbol("x", positive = True) # definicja symbolu matematycznego w SymPy
```

```python
X = Exponential("x",a) # definicja symbolicznego rozkłądu expotencjalnego  w SymPy
```

```python
#z = Symbol("x")
D = density(X)(x) # utworzenie obiektu reprezentującego gęstość powyższego rozkładu
D # wypisz wyrażenie reprezentujące gęstość
```

```python
CDF = cdf(X)(x)
CDF
```

```python
[E(X), variance(X)] # obliczenie wartości oczekiwanej i wariancji
```

### Kombinatoryka symboliczna

```python
from sympy import *
init_printing()
```

```python
n = Symbol('n', integer=True)
f1 = factorial(n+3)
f2 = factorial(n)

print("upraszczanie silni:")
simplify(f1/f2)
```

Permutacje i kombinacje:
$$P_{k,n} = \frac{n!}{(n-k)!}$$

$$C_{k,n} = \frac{n!}{k! (n-k)!}$$


```python
# przykład funkcji obliczającej permutację
def P(k,n):
    return factorial(n)/factorial(n-k)

# przykład funkcji obliczającej kombinację
def C(k,n):
    return factorial(n)/(factorial(k)*factorial(n-k))
```

```python
C(Symbol("k"),Symbol("n")) # symbolic use
```

```python
C(2,10) # aplikacja do liczb
```

```python
C(Symbol("k"),10) 
```

```python
C(6,49) # liczba możliwych wyników w lotto
```
