Zmienna losowa 𝑋 ma rozkład normalny 𝑁(125,15). W języku Python:
 1. Utwórz instancję tego rozkładu;
 2. Narysuj gęstość rozkładu prawdopodobieństwa i dystrybuantę w zakresie $[\mu-4\sigma;\mu+4\sigma]$;
 3. Oblicz prawdopodobieństwo, że zmienna losowa osiągnie wartość z zakresu 130 do 145;
 4. Oblicz wartość oczekiwaną tej zmiennej losowej;
 5. Oblicz odchylenie standardowe tej zmiennej losowej;
 6. Oblicz kwantyl rzędu 0.25 tej zmiennej losowej.

```python
from scipy import stats as st # import podstawowej biblioteki z obiektami rozkładów prawdopodobieństwa
import numpy as np # wymagane do manipulacji tablicami numpy
import matplotlib.pyplot as plt #biblioteki wymagane do wykresu
```

```python
m = 125 # średnia
s = 15 # odchylenie standardowe
NormalDistribution = st.norm(m,s) # utworzenie rozkładu
```

```python
NormalDistribution
```

```python
x = np.linspace(m-4*s,m+4*s,100) # wektor wartości x
f = NormalDistribution.pdf(x) # wektor wartości gęstości rozkładu prawdopodobieństwa
F = NormalDistribution.cdf(x) # obliczenie wartości dystrybuanty

plt.plot(x,f,"blue",x,F,"red")
plt.legend(["f(x) - gęstość","F(x) - dystrybuanta"])
pass
```

```python
E = NormalDistribution.expect()
S = NormalDistribution.std()
q025 = NormalDistribution.ppf(0.25)
[E,S,q025]
```
