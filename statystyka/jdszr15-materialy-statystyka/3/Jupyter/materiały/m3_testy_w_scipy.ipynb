Moduł Pythona [scipy](https://docs.scipy.org/doc/scipy/reference/stats.html) oferuje znaczną ilość ciekawych funkcji statystycznych np. rozkłady i testy statystyczne.
Innym ciekawym modułem jest moduł [StatsModels](https://www.statsmodels.org/stable/index.html)

```python
import numpy as np
import scipy as sp
import scipy.stats as st
import matplotlib.pyplot as plt
%matplotlib inline
```

**Wygenerowanie próby pochodzącej z rozkładu normalnego**

```python
x = st.norm(5,1).rvs(100)
```

**Sprawdzenie normalności próby - test [Shapiro-Wilka](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.shapiro.html#scipy.stats.shapiro)**

```python
st.shapiro(x)
```

**Tzw. wykres prawdopodobieństwa normalnego [probplot](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.probplot.html#scipy.stats.probplot)**

```python
st.probplot(x, plot=plt)
pass
```

**test [$\chi^2$ dopasowania](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.chisquare.html#scipy.stats.chisquare)**

Porównanie wartości otrzymanych na podstawie próby z wartościami rozkładu teoretycznego

```python
czestosci_obserwowane = np.array([100455,42357,27930,23222,18535,14081,13771,12740,14912])
czestosci_teoretyczne = np.log10(1+1/np.array(range(1,10))) * np.sum(czestosci_obserwowane)
st.chisquare(czestosci_obserwowane, f_exp=czestosci_teoretyczne)
```
