W zeszycie przedstawiono przykładową funkcję obliczającą statystyki testu dla średniej

Funkcja korzysta tylko z rozkładu T-Studenta, gdyż dla dużej próby jest on zbieżny do rozkładu normalnego.

```python
import numpy as np
import scipy as sp
import scipy.stats as st
```

```python
def TTest(x, mu0=0, AlternativeHypothesis="Unequal"):
    N = len(x)
    df = N-1    
    StudentTDistribution = st.t(df)
    
    t = (st.tmean(x)-mu0)*np.sqrt(N)/st.tstd(x)
    CDFvalue = StudentTDistribution.cdf(t)
    
    if AlternativeHypothesis == "Less":
        pvalue = CDFvalue
    elif AlternativeHypothesis == "Unequal":  
        pvalue = 2.*np.min([CDFvalue,1.-CDFvalue])
    elif AlternativeHypothesis == "Greater":
        pvalue = 1.-CDFvalue
    else:
        raise ValueError('AlternativeHypothesis argument can take value: Less, Unequal or Greater')
    return pvalue  
```

```python
x = np.array([5,4,3,1,7])
[TTest(x,AlternativeHypothesis="Less"),
 TTest(x,AlternativeHypothesis="Unequal"),
 TTest(x,AlternativeHypothesis="Greater")]
```
