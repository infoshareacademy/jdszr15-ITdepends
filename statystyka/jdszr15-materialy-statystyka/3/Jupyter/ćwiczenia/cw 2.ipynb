Dany jest zbiór informacji o aplikacjach z Google Play. Przyjmując poziom ufności $1−\alpha=0,99$, określ przedział ufności dla proporcji liczby aplikacji, które zostały zainstalowane 1 000 000 000+ razy.

```python
import scipy.stats as st
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("googleplaystore.csv") # import danych do obiektu DataFrame
print(df.columns) # wypisanie nazw kolumn
df.head() # wypisanie kilku początkowych wierszy danych
df['Installs'].unique() # wypisanie unikalnych wartości z kolumny "Installs"
```

**Wybór danych, dla których kolumna to Installs**

```python
df = df['Installs']
df.head()
```

**Czy DataFrame zawiera obserwacje NaN?**

```python
print(not df.count() == len(df) )
```

```python
k = sum(df == '1,000,000,000+') # obliczenie liczby aplikacji, które zainstalowano 1.000.000.000+ razy
n = df.count() # obliczenie liczby wszystkich aplikacji (które figurują w tym zbiorze danych)

p = k/n
a = 0.01

NormalDistribution = st.norm() # utworzenie instancji obiektu rozkładu normalnego standardyzowanego
ua = NormalDistribution.ppf(1-a/2) # obliczenie prawego kwantyla rozkładu normalnego standardyzowanego

CIp = np.array( [p - ua*np.sqrt(p*(1-p)/n), p + ua*np.sqrt(p*(1-p)/n)] ) # obliczenie przedziałów ufności

print("Przedział ufności pokrywający wartość proporcji w populacji generalnej z p={}:".format(1-a))
print(CIp)
```

#### Użycie funkcji bootstrap

W celu wykorzystania funkcji [`bootstrap`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.bootstrap.html) do obliczenia przedziałów ufności proporcji konieczne jest zdefiniowanie nowej funkcji.

```python
def proportion(x):
    k = sum(x)
    n = len(x)
    return k/n
```

```python
x = (df.to_numpy() == '1,000,000,000+') # True jeśli spełniony jest warunek, False w przeciwnym razie
st.bootstrap((x,),proportion,confidence_level=1-a,vectorized=False)
```
