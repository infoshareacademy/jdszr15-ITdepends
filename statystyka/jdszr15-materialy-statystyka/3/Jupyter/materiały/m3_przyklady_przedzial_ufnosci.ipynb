**Przykład 1**

Badano średnią kwotę na jaką klienci pewnego marketu dokonują zakupów w niedzielę handlową. 
Zanotowano kwotę zakupów $n=130$ klientów. Otrzymano średnią $\bar{x}=124$ PLN przy odchyleniu standardowym $𝑠 = 50$ PLN.  Zbuduj przedział ufności dla średniej ceny zakupów przy poziomie ufności $1−\alpha=0,95$.

```python
import scipy.stats as st
import numpy as np
```

**Obliczenie przedziałow ufności**

```python
# wprowadzenie danych
n = 130
m = 124
s = 50

a = 0.05

NormalDistribution = st.norm() # utworzenie instancji obiektu rozkładu normalnego standardyzowanego

ua = NormalDistribution.ppf(1-a/2) # obliczenie prawego kwantyla rozkładu normalnego standardyzowanego
CIm = np.array( [m - ua*s/np.sqrt(n), m + ua*s/np.sqrt(n)] ) # obliczenie granic przedziałów ufności

print("Przedział ufności pokrywający średnią wartości w populacji generalnej z p={}:".format(1-a))
print(CIm)
```

**Przykład 2**

W sklepie internetowym zbadano liczbę transakcji z jednego dnia. Dokonano $n=5000$ zakupów, z czego $200$ było wynikiem przekierowania (kliknięcia w reklamę na stronie internetowej). Na poziomie ufności $1−\alpha=0,99$ określ przedział ufności dla proporcji.

```python
n = 5000
k = 200
p = k/n
a = 0.01

NormalDistribution = st.norm() # utworzenie instancji obiektu rozkładu normalnego standardyzowanego
ua = NormalDistribution.ppf(1-a/2) # obliczenie prawego kwantyla rozkładu normalnego standardyzowanego

CIp = np.array( [p - ua*np.sqrt(p*(1-p)/n), p + ua*np.sqrt(p*(1-p)/n)] )

print("Przedział ufności pokrywający wartość proporcji w populacji generalnej z p={}:".format(1-a))
print(CIp)
```
