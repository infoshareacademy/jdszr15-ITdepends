Określ przedział ufności dla stopnia konwersji (conversion rate) użytkowników aktualnej strony internetowej (old_website). Przyjmij poziom ufności $1−\alpha = 0,999$.

```python
import scipy.stats as st
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("ab_data.csv") # import danych do obiektu DataFrame
print(df.columns) # wypisanie nazw kolumn
df.head() # wypisanie kilku początkowych wierszy danych
```

```python
df = df.loc[df['landing_page']=="old_page", 'converted']
df.head()
```

**Czy DataFrame zawiera obserwacje NaN?**

```python
print(not df.count() == len(df) )
```

**Wykonanie obliczeń**

```python
k = sum(df == 1) # obliczenie liczby użytkowników, którzy dokonali konwersji
n = df.count() # obliczenie liczby wszystkich badanych użytkowników

p = k/n # obliczenie proporcji w próbie
a = 0.001 # poziom istotności

NormalDistribution = st.norm() # utworzenie instancji obiektu rozkładu normalnego standardyzowanego
ua = NormalDistribution.ppf(1-a/2) # obliczenie prawego kwantyla rozkładu normalnego standardyzowanego

CIp = np.array( [p - ua*np.sqrt(p*(1-p)/n), p + ua*np.sqrt(p*(1-p)/n)] ) # obliczenie przedziałów ufności

print("Przedział ufności pokrywający wartość proporcji w populacji generalnej z p={}:".format(1-a))
print(CIp)
```

```python
def proportion(x,axis=None):
    k = np.sum(x)
    n = len(x)
    return k/n

x = (df.to_numpy() == 1) # True jeśli spełniony jest warunek, False w przeciwnym razie
st.bootstrap((x,),proportion,confidence_level=1-a,vectorized=False,batch=1000,n_resamples=999)
```
