Przykład przeprowadzenia testu $\chi^2$ niezależności

```python
import numpy as np
import scipy.stats as st
```

Dane: tabela obserwowanych częstości

```python
nij = np.array([[100, 140, 100],
                [ 10,  30,  20]])
```

Obliczenie sum częstości wystąpienia zdarzeń, czli sum wierszy i kolumn

```python
W = np.sum(nij,axis = 1) # suma wierszy
K = np.sum(nij,axis = 0) # suma kolumn
N = np.sum(nij) # suma wszystkich elementów
```

Obliczenie prawdopodobieństw odpowiadających wierszom i kolumnom

```python
pW = W/N
pK = K/N
```

Mnożąc prawdopodobieństwa zdarzeń otrzymuje się teoretyczne prawdopodobieństwa iloczynów zdarzeń, tzn. takie, dla których zdarzenia byłyby niezależne. Następnie na ich podstawie oblicza się częstości teoretyczne.

```python
pijT = np.tensordot(pW,pK,axes=0)
nT = pijT * N
print(nT)
```

**Obliczenie statystyki testowej i p-wartości**

```python
chi2t = np.sum((nT-nij)**2/nT)

df = (len(W)-1)*(len(K)-1)
rozkladChi2 = st.chi2(df)

pvalue = 1-rozkladChi2.cdf(chi2t)

print("Statystyka testowa chi2 = {}; pvalue = {}".format(chi2t,pvalue))
```

P-wartość jest większa niż przyjęty poziom istotności $\alpha=0,05$ zatem brak jest podstaw do odrzucenia hipotezy zerowej mówiącej o niezależności zdarzeń. Można uznać, że zdarzenia są niezależne, czyli metoda produkcji nie ma wpływu na ilosć sztuk wadliwych.
