## Użycie obiektu dict() Pythona do obliczeń prawdopodobieństwa

Niech $\Omega$={($x$,$y$):0<$x$<1,0<$y$<1}; $A_1$, $A_2$, ..., $A_7$ będą zdarzeniami zaznaczonymi na rysunku. Oznaczmy: $A$=$A_1$∪$A_2$∪$A_3$; $B$=$A_1$∪$A_4$∪$A_7$; $C$=$A_3$∪$A_6$∪$A_7$. Zbadać, czy zdarzenia $A$, $B$, $C$ są niezależne oraz czy są parami niezależne.
![$\Omega$](A7.png)

```python
from fractions import Fraction as f #import obiektu reprezentującego ułamki

def PE(s:set, P:dict): # funkcja obliczająca prawdopodobieństwo zdarzenia złożonego
    if s == set():
        return 0
    else:
        p = 0
        for e in s:
            p += P[e]
        return p


P = dict() # słownik do przechowywania prawdopodobieństw w wygodny sposób
for e in range(1,8):
    E = "A"+str(e)
    P[E] = f(1,9) # użycie obiektu Fraction; wszystkie prawdopodobieństwa wynoszą 1/9

print(P)    

#definiowanie zdarzeń poprzez litery
A = {"A1","A2","A3"}
B = {"A1","A4","A7"}
C = {"A3","A6","A7"}

#obliczenie prawdopodobieństw zdarzeń złożonych
PA = PE(A,P)
PB = PE(B,P)
PC = PE(C,P)

print("P(A) = {}, P(B) = {}, P(C) = {}".format(PA,PB,PC))

intersectionABC = A & B & C # iloczyn zbiorów A, B i C
PintersectionABC = PE(intersectionABC,P) # prawdopodobieństwo powyższego iloczynu zdarzeń
P_ABC = PE(A,P) * PE(B,P) * PE(C,P) # iloczyn prawdopodobieństw zdarzeń A, B i C
print("Iloczyn zbiorów A, B oraz C = {}".format(intersectionABC))
print("P(A & B & C) = {} {} P(A)*P(B)*P(C) = {}".format(PintersectionABC,"równe" if PintersectionABC==P_ABC else "nie równe",P_ABC))

P_AB = PE(A,P)*PE(B,P) # iloczyn prawdopodobieństw zdarzeń A i B
PintersectionAB = PE(A & B,P) # prawdopodobieństwo iloczynu zdarzeń A i B
P_AC = PE(A,P)*PE(C,P)  # iloczyn prawdopodobieństw zdarzeń A i C
PintersectionAC = PE(A & C,P)# prawdopodobieństwo iloczynu zdarzeń A i C
P_BC = PE(B,P)*PE(C,P)  # iloczyn prawdopodobieństw zdarzeń B i C
PintersectionBC = PE(B & C,P) # prawdopodobieństwo iloczynu zdarzeń B i C

print("P(A & B) = {} {} P(A)*P(B) = {}".format(PintersectionAB,"równe" if PintersectionAB==P_AB else "nie równe",P_AB))
print("P(A & C) = {} {} P(A)*P(C) = {}".format(PintersectionAC,"równe" if PintersectionAC==P_AC else "nie równe",P_AC))
print("P(B & C) = {} {} P(B)*P(C) = {}".format(PintersectionAC,"równe" if PintersectionBC==P_BC else "nie równe",P_BC))
```

**Prawdopodobieństwo zupełne**

W laboratorium znajduje się piętnaście automatycznych mierników jakości powietrza. Trzy nowego typu (prawdopodobieństwo prawidłowego pomiaru wynosi $p_1=0,98$) i reszta starego typu (prawdopodobieństwo prawidłowego pomiaru wynosi $p_2=0,94$). Pomiar przeprowadzono jednym (losowo wybranym) urządzeniem. Znaleźć prawdopodobieństwo tego, że pomiar będzie prawidłowy. 


$$P(A) = P(A|B_1) \cdot P(B_1) + P(A|B_2) \cdot P(B_2)$$

```python
from fractions import Fraction as f #import obiektu reprezentującego ułamki

P_AB1 = f(3,15)
P_B1 = f(98,100)
P_AB2 = f(12,15)
P_B2 = f(94,100)

PA = P_AB1 * P_B1 + P_AB2 * P_B2
print("{} = {}".format(PA,float(PA)))
```

**Twierdzenie Bayesa**

Jakie jest prawdopodobieństwo, że przyczyną kaszlu jest grypa, jeśli 1% populacji ma grypę, a 5% populacji ma kaszel i 90% ludzi mających grypę kaszle.

$$P(GRYPA|KASZEL) = \frac{P(KASZEL|GRYPA) \cdot P(GRYPA)}{P(KASZEL)}$$

```python
from fractions import Fraction as f #import obiektu reprezentującego ułamki

P_grypa = f(1,100) # prawdopodobieństwo, że losowo wybrana osoba ma grypę 1%
P_kaszel = f(5,100) # prawdopodobieństwo, że losowo wybrana osoba ma kaszel
P_kaszel_grypa = f(90,100) # prawdopodobieństwo, że losowo wybrana osoba ma kaszel, jeśli ma grypę

P_grypa_powoduje_kaszel = P_kaszel_grypa * P_grypa / P_kaszel
print("{} = {}".format(P_grypa_powoduje_kaszel,float(P_grypa_powoduje_kaszel) ))
```
