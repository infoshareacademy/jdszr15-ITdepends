**W eksperymencie bada ilość wydzielonego produktu w wyniku reakcji chemicznej. Wykonano $n=60$ niezależnych doświadczeń i otrzymano z nich średnią  $\bar{x}=46$ g oraz odchylenie $s=13$ g. Przyjmując współczynnik ufności $1-\alpha=0,99$ oszacować metodą przedziałową średnią wagę wydzielonej substancji.**

```python
import numpy as np
import scipy.stats as st
```

```python
# Utworzenie zmiennych przechowujących dane:
x=46
s=13
a=1-0.99
n=60
```

```python
# Utworzenie obiektu reprezentującego rozkład normalny standardyzowany:
NormalDistribution = st.norm(0,1)
```

Odczytanie kwantyli (funkcja ppf) rozkładu normalnego standardyzowanego dla wartości $\frac{\alpha}{2}$ oraz $1-\frac{\alpha}{2}$:

```python
u=NormalDistribution.ppf([a/2,1-a/2])
u
```

Obliczenie przedziałów ufności. Wstawienie do wzoru:
$$CI_{\mu} = \left( \bar{x} + u_{\frac{\alpha}{2}}\frac{s}{\sqrt{n}}; \bar{x} + u_{1-\frac{\alpha}{2}}\frac{s}{\sqrt{n}}\right)$$

```python
CIm = x + u*s/np.sqrt(n)
print(CIm)
```

**Dla danych z zadania powyżej. zbuduj przedział ufności dla odchylenia standardowego.**

```python
# Utworzenie obiektu reprezentującego rozkład chi kwadrat z n-1 stopniami swobody
ChiSquareDistribution = st.chi2(n-1)
```

```python
# obliczenie kwantyli tego rozkładu dla danych wartości a/2 i 1-a/2
q=ChiSquareDistribution.ppf([a/2,1-a/2])
q
```

Obliczenie przedziałów ufności wg wzoru (zwróć uwagę, na kolejność kwantyli we wzorze)
$$CI_\sigma = \left(\sqrt{\frac{(n-1)s^2}{\chi_{1-\frac{\alpha}{2}}^2}} ; \sqrt{\frac{(n-1)s^2}{\chi_{\frac{\alpha}{2}}^2}} \right)$$

```python
CIs = np.sqrt((n-1)*s**2/q[::-1])
print(CIs)
```

**W ankiecie wzięło udział $n=1000$ respondentów. Pytano ich czy palą regularnie. 20% z nich odpowiedziało „tak”. Znajdź przedziały ufności dla proporcji osób palących w populacji generalnej. przyjmij poziom ufności
$1-\alpha = 0,99$.**

```python
n=1000
p=20/100
a=1-0.99


# Utworzenie obiektu reprezentującego rozkład normalny standardyzowany:
NormalDistribution = st.norm(0,1)
```

```python
u=NormalDistribution.ppf([a/2,1-a/2]) #odczytanie kwantyli
u
```

```python
CIp = p+u*np.sqrt(p*(1-p)/n)
print(CIp)
```
