## Prezentacja możliwości moduł [SymPy](https://www.sympy.org/en/index.html)

Moduł do **sympy** służy do symbolicznej manipulacji obiektami matematycznymi

```python
from IPython.display import display # ładny wydruk obiektów
from sympy import *
init_printing()
```

Aby wykorzystać zmienną jako symbol należy utworzyć obiekt biblioteki Sympy

```python
x = Symbol("x")
display(x)
y = Symbol("y")
display(y)
```

Dostępne są pewne symbole specjalne

```python
a = Symbol("alpha")
a
```

Następnie używając symboli możliwe jest budowanie **wyrażeń**

```python
f = a*exp(-a*x)
f
```

Funkcja **Subs** umożliwia podstawianie wartości za symbol

```python
Subs(f,x,y)
```

Aby wykonać podstawienie należy użyć funkcji **.doit()** 

```python
Subs(f,x,y).doit()
```

```python
f
```

.doit() nie zmienia oryginalnego obiektu, zatem aby zachować wynik operacjinależy podstawić go do zmiennej

```python
g = Subs(f,x,y).doit()
g
```

Wyrażeniami operuje się identycznie jak symbolami

```python
h=f+g
h
```

```python
expand((x+y)**3) # rozwinięcie wzoru skróconego mnożenia
```

```python
i = integrate(cos(x)**2,x) #całkowanie
i
```

```python
d = diff(i,x) #różniczkowanie
display(d)
simplify(d) #uproszczenie
```

```python
eq = Eq(2/pi*atan(exp(x)),y)
solve(eq,x)
```

```python
limit(atan(exp(x)),x,+oo) # obliczenie granicy funkcji
```

```python
plot(sin(x),(x,0,2*pi))
```

```python
A = Matrix([[1, -x], [3, x**2]])
display(A)
det(A)
```
